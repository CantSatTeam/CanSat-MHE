```python
# view depth tiff map
import imageio.v3 as iio
import numpy as np
import matplotlib.pyplot as plt

depth = iio.imread("depth_map.tiff").astype(np.float32)

# handle invalids if needed
depth[depth <= 0] = np.nan

plt.imshow(depth)
plt.colorbar()
plt.title("Depth Map")
plt.axis("off")
plt.show()
```
